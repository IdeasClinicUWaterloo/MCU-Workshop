# Exercise 2: Additive Synthesis & Envelopes

## Part 1: Additive Synthesis & Harmony

Delete any existing tracks in Audacity. On two **different** tracks, generate:
* Track 1: Sine wave with frequency **523 Hz** ($C_5$), amplitude **0.5**, duration **2 seconds**
* Track 2: Sine wave with frequency **784 Hz** ($G_5$), amplitude **0.5**, duration **2 seconds**

First space them out to listen separately, then align them to play simultaneously. Because of **superimposition**, the combined wave you hear is the mathematical sum of each pure tone—producing a full, harmonized musical fifth rather than the wobbling beats from Exercise 1.

## Part 2: Synthesizer Envelopes (Non-Constant Amplitude & Frequency)

In Exercise 1, beats created a wave with a time-varying amplitude:

$$2A\cos\left(\left(\frac{\omega_1 - \omega_2}{2}\right)t\right)\sin\left(\left(\frac{\omega_1 + \omega_2}{2}\right)t\right)$$

Synthesizers replace that periodic envelope with deliberate functions—such as a decaying exponential $16e^{-6t}\sin(\omega t)$—so a note's volume fades out naturally after a key is pressed:

![Non-constant Amplitude Envelope](../../assets/envelope_amplitude_function.png)

![Exponential Decay Waveform](../../assets/envelope_exponential_decay.png)

We can apply the exact same principle to **frequency**. Because angular frequency $\omega$ is inversely proportional to period ($T = \frac{2\pi}{\omega}$), defining frequency with a decaying exponential causes a sharp drop in pitch—which is how electronic synthesizers synthesize kick drums and percussion:

![Non-constant Frequency Drum Envelope](../../assets/envelope_frequency_drum.png)

Conversely, modulating frequency with a slow sinusoid produces a rising and falling siren effect:

![Sinusoid Frequency Modulation](../../assets/envelope_frequency_siren.png)
