# Exercise 1: Sound Waves – Pitch, Amplitude & Beats

In this section, you will observe how frequency, amplitude, and wave interference affect how a sound wave looks and sounds.

* **Frequency** refers to the number of oscillations per second, measured in **hertz (Hz)**. We perceive frequency as **pitch**.
* **Amplitude** denotes the maximum value of an oscillation measured from equilibrium. For sound waves, amplitude corresponds to **volume** (measured on the logarithmic decibel scale).

## Part 1: Generating a Pure Tone

In an empty project in Audacity, select **Generate > Tone** from the top toolbar with the following settings:
* Waveform: **Sine**
* Frequency: **440 Hz**
* Amplitude: **0.8**
* Duration: **2 seconds**

![Audacity Tone Generator Settings](../../assets/audacity_tone_generator.png)

Zoom in on the generated waveform until you can clearly see individual cycles of the sine wave. Recall the general form of a sine wave:

$$y = A\sin[B(x - C)] + D$$

### Questions
1. **What is the theoretical period ($T = \frac{1}{f}$) of the 440 Hz wave you created?**
2. **Given that the period of a sine function is $\frac{2\pi}{B}$, find $B$ and write the full equation $y = A\sin(Bx)$ for this wave.**

## Part 2: Beats – Time-Varying Amplitudes

Delete your existing Audacity tracks. On two **separate** tracks, generate:
* Track 1: Sine wave with Frequency **300 Hz**, Amplitude **1**, Duration **2 seconds**
* Track 2: Sine wave with Frequency **305 Hz**, Amplitude **1**, Duration **2 seconds**

Play both tracks simultaneously. When two sound waves with slightly different frequencies overlap, their constructive and destructive interference creates periodic volume pulses called **Beats**:

$$f(t) = A\sin(300 \cdot 2\pi t) + A\sin(305 \cdot 2\pi t)$$

### Challenge
Count how many volume pulses (beats) occur per second when playing the **300 Hz** and **305 Hz** tones together. How does this pulse rate relate to the difference between the two frequencies?
