# Exercise 3: Hardware Diagnostics & Connection Check

Before programming musical features, we need to verify that every potentiometer and pushbutton is properly soldered and communicating with the **Seeeduino Nano**.

## Running `PCB_CONNECTION_CHECKER.ino`

1. **[Download PCB_CONNECTION_CHECKER.zip](https://github.com)**, unzip it, and open `PCB_CONNECTION_CHECKER.ino` in the Arduino IDE. If prompted to create a sketch folder, click **OK**.
2. Connect your Seeeduino Nano to your computer using the USB-C to USB-A cable.
3. Open the **Select Board** dropdown, choose **Select Other Board and Port**, pick **Seeeduino Nano**, select your active COM/USB port, and click **OK**.
4. Click **Verify** (checkmark) to confirm the code compiles, then click **Upload** (right arrow).
5. Open the **Serial Monitor** (`Ctrl + Shift + M` on Windows/Linux or `Cmd + Shift + M` on macOS).

### Expected Serial Monitor Behavior
* **Potentiometers:** Turning each dial clockwise should smoothly increase its reading; turning counterclockwise should decrease it.
* **Pushbuttons:** Because the pins use `INPUT_PULLUP`, each button should read **`1` (HIGH)** when untouched and **`0` (LOW)** when pressed.

## Troubleshooting

* **Serial output scrolling too fast:** Add a brief delay (e.g., `delay(500);` or `delay(1000);`) at the bottom of `loop()` so you can read individual pin values clearly.
* **Unresponsive pin or stuck value:** Double-check your solder joints on that component and consult a workshop facilitator before moving on.
