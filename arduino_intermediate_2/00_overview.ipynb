# Synthesizer Architecture & Overview

In many modern songs, we hear sounds that are not recognizable as traditional instruments such as the guitar, piano, violin, or drums. These sounds are created by electronic musical instruments called **synthesizers**, or a software imitation thereof.

Analog and digital synthesizers typically create sounds by generating waveforms through methods including subtractive synthesis, additive synthesis, and frequency modulation (FM) synthesis. These sounds are altered by:
* **Filters:** Cut or boost specific frequencies.
* **Envelopes (ADSR):** Control articulation, or how notes begin, sustain, and fade out.
* **Low-Frequency Oscillators (LFOs):** Modulate parameters such as pitch, volume, or filter cutoff over time.

## Synthesizer Modes

The synthesizer you will be programming has two modes that you can switch between by pressing a mode button. Each mode reassigns the PCB potentiometers:

* **LFO State Potentiometers:**
  * Waveform (sine, triangle, saw, and square waves)
  * Frequency / Pitch
  * Cutoff / Intensity
  * LFO Cutoff Rate
  * Resonance (the vibrating quality of the sound)
  * Amplitude / Volume

* **ADSR State Potentiometers:**
  * Attack
  * Decay
  * Sustain
  * Release
