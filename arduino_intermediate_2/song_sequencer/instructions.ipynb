# Final Challenge: Custom Song Sequencer

The sequencer firmware can automatically play multi-note melodies by stepping through parallel `NOTES` and `TIMES` arrays.

## Part 1: Exploring Preset Melodies

1. Unzip `SYNTH_SEQUENCER.zip` and open `SYNTH_SEQUENCER.ino` in the Arduino IDE.
2. Select **Seeeduino Nano** and your port, then upload the code.
3. Press the buttons on your PCB to trigger the built-in sequences. While a sequence plays, adjust the **Attack**, **Decay**, **Sustain**, and **Resonance** potentiometers to sculpt the instrument timbre to match the original song.

## Part 2: How `NOTES` and `TIMES` Arrays Work

* `NOTES_X[]` stores the sequence of pitches to be played.
* `TIMES_X[]` stores the corresponding duration for each note.
* **Crucial Rule:** The integer inside the square brackets `[N]` must match the exact number of comma-separated elements inside the curly braces `{...}`, and both arrays must have the exact same length.

![NOTES and TIMES Arrays](../assets/sequencer_notes_times_arrays.png)

## Part 3: Program Your Own Song

1. **Transcribe a Melody:** Pick a recognizable song (e.g., *Mary Had a Little Lamb*, *Tetris*, or a favorite theme) and overwrite `NOTES_3` and `TIMES_3` in `SYNTH_SEQUENCER.ino`.
2. **Update Array Length:** Remember to update the element count inside the square brackets `[]` (circled below) to match your new melody's length:

![Array Size Initialization](../assets/sequencer_array_size.png)

3. **Enable Looping (Optional):** To repeat a sequence continuously, set `sequence_looping_X = true` (where `X` is your sequence number). When looping is enabled, you **must** set the very first element of `NOTES_X` to `REST` and the very first element of `TIMES_X` to `0` for the loop to reset properly.
4. **Swap the Oscillator Engine:** Open `internal.cpp` in the project folder. Comment out (`//`) the active oscillator line and uncomment the oscillator waveform you want to use:

![Oscillator Selection in internal.cpp](../assets/sequencer_oscillator_selection.png)

## Troubleshooting

* **Dynamic Memory Exceeds 100% Error:** If the compiler warns that global variables exceed 100% of dynamic memory, shorten your `NOTES_3` and `TIMES_3` arrays to free up SRAM.
